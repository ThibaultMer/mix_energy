```python
import requests

```

```python
url = "https://odre.opendatasoft.com/api/explore/v2.1/catalog/datasets/eco2mix-national-cons-def/exports/csv"
response = requests.get(url)
```

```python
import os
fichier = "../../data/nat_consolidee.csv"
if response.status_code == 200:
    with open(fichier, "wb") as f:
        f.write(response.content)
    print(f"Fichier enregistré : {fichier}")
else:
    print(f"Erreur : {response.status_code}")
```

```python
try:
    # Requête GET
    response = requests.get(url)

    # Vérifier si la requête a réussi (code 200)
    if response.status_code == 200:
        data = response.json()  # Convertir la réponse en JSON
        print("Données reçues :")
        print(data)
    else:
        print(f"Erreur : {response.status_code}")

except requests.exceptions.RequestException as e:
    print(f"Une erreur s'est produite : {e}")
```

```python
import requests

url = "https://api.exemple.com/data"

response = requests.get(url)

if response.status_code == 200:
    data = response.json()
    
    print("=== Résultat de l'API ===")
    print(data)  # affiche tout le JSON

    # Optionnel : affichage plus propre
    print("\n=== Affichage formaté ===")
    for cle, valeur in data.items():
        print(f"{cle} : {valeur}")

else:
    print(f"Erreur : {response.status_code}")
```
