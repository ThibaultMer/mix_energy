```python
import os
import pandas as pd
import requests
import io

from mix_energy import get_logger


def get_base_carbone():
    """
    Importe les données de la base carbone de l'Ademe depuis un fichier csv depuis le site de l'Ademe,
    nettoie les données et importe le csv dans dans le bucket GCP.
    """
    logger = get_logger()
    url = os.getenv("BASE_CARBONE_URL")
    df = pd.DataFrame()
    if not url:
        logger.error("BASE_CARBONE_URL n'est pas définie")
        
    else:
        response = requests.get(url)
        response.raise_for_status()
        logger.info(f"URL: {response.url[:50]}...")
        df = pd.read_csv(io.BytesIO(response.content), sep=";", encoding="cp1252")
        
    return df
```
